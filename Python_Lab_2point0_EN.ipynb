You can print several different things at once. The next piece of code prints text, a number, and text again. The "units" of information are separated by commas.

````python
print("I am", 39, "years old")
````

Information can be stored in variables. In this case let us call the value 39 simply `age`. Change the value of `age` to your own age, create a new variable `name` and assign your name to it. Print the information in the format "My name is `name` and I am `age` years old".

````python
age = 39
print ("I am", age, "years old")
````

In Python, symbols that are not inside quotation marks are treated not as information but as functions or variables that carry a second meaning. For example, `"age"` is treated as a word, while `age` is treated as a reference/variable pointing to a box holding information.

````python
age = 39
print (age)
print ("age")
````

Information can be given to a program in several ways: written down as a variable `a = 2`, read from a file `a = open("file.txt")`, downloaded from the internet, or typed in by the user. For the last one we use the function `input()`. As the function's argument you can give a piece of text - a prompt shown while the text is being entered: `input("What do we want to ask")`. The value that is read in is always text, even when a number is typed.

````python
day = input("Enter what day of the week it is today: ")
print ("Today is", day)
````

Most functions, once used, return some value or object to us. `input()` is no exception. This function returns the value we type on the keyboard, and once we have a value, we have to "put it in a box": `value = input()`.

If we simply write `input()`, we do get the value, but without a variable (a reference to it) it is worthless and unusable. In the example, pay attention to the function `int()` - all data read with `input()` is text (`string`), so to square the entered text (`"7"`) it must first be converted from text into a number (`string` to `integer`).

ChatGPT:
The decimal number 7 is 111 in binary. As a string, "7" represents the character '7', which is different from its numerical binary equivalent. If you were to convert the character '7' to binary using the ASCII encoding, its decimal value is 55, which in binary is 110111.


````python
input ("Enter the first number: ")                   # A number is entered, but without assigning it to a variable we cannot use it
second_num = int (input ("Enter the second number: "))  # This number is entered and can be used in calculations, e.g. num_squared = second_num**2
print (second_num**2)
````
