In Python all data are objects. This means that each of them is described by certain rules that determine what can be done with them. Not everything you can do with numbers can be done with words, and the other way round. For example, the letters of a piece of text can be turned into capitals, `"seven".upper()`, but you will not manage the same with a number, `7.upper()`.

````python
# The type of a variable can be checked with the function type()
x = "name"
y = x
z = 7
w = 12.3

print (type(x))
print (type(y))
print (type(z))
print (type(w))
````

In the example above, change the variable `z` to `float` type and change the variable `w` to `string` type.

We can convert variables between different types using the functions `int()`, `float()` and `str()`.

````python
number = 123
text = str(number)
number_2 = float(text)

print (type(number))
print (type(text))
print (type(number_2))
````

Text (a string) is a composite object - made up of individual characters. They can be indexed.

````python
text = "Third programming seminar in 2026"

first_letter = text[0]

print (first_letter)

# You can also select more than one character by using indexing

first_word = text[0:5]

print (first_word)
````

Modify the code above so that it cuts out "2026". Convert it to a number and divide it by 5.

A list is also an object. The characters of a text can easily be converted into a list with the function `list()`. This function turns every character of the text into a separate member of the list. If we want to split at particular characters, we can use `string.split()` - where string is some piece of text.

````python
text = "To the mind that is still, the whole universe surrenders."

items = list(text)
items2 = text.split(",")   # We split the text into parts at the comma
print (items2)
````

Modify the code above so that the list contains the individual words. You will need two steps for this. First remove the comma from the text with the `string.replace()` method, and then use `string.split()` to split at the spaces.

A list is also a composite object (made up of members), so we can index it. In the code example below we split the sentence at the comma, which means we have two members (before and after the comma), and by indexing we pick the first part of the sentence with index `[0]` - in programming the first member is number 0.

````python
text = "To the mind that is still, the whole universe surrenders."

items = text.split(",")

print (items[0])
````
