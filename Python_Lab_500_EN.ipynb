**The four parts of a while loop:**

1. **PREPARE** the variable the condition looks at - **before** the loop
2. **CONDITION** - checked **before** every pass
3. **ACTIONS** - what the loop does
4. **CHANGE** - without it the loop never stops

> **If a loop does not stop:** Colab - the ■ button beside the cell (or Runtime → Interrupt execution, or Ctrl+M, then I). Do not close the browser.

A `while` loop is an open loop (the number of passes is not fixed in advance): it repeats as long as some condition holds.

````python
# Create the counter value
count = 1

# The while loop checks the condition before every pass
while count <= 5:
    print(f"Current count is: {count}")

    # Change the counter value
    count = count + 1

# This part runs once the while condition no longer holds
print("The loop has finished counting.")
````

Dictionaries, used well, are a very convenient way to store information. Information is stored as key:value pairs.

````python
# 1. Create a dictionary where the keys are names (strings)
#    and the values are test scores (integers).
student_grades = {"Alice": 88, "Bob": 92, "Charlie": 76}

print("--- Initial Grades ---")
print(student_grades)

# 2. Values are reached through their keys
bobs_score = student_grades["Bob"]
print(f"\nBob's score is: {bobs_score}")

# 3. A new key:value pair can be added
student_grades["Diana"] = 95
print(f"\nAfter adding Diana: {student_grades}")

# 4. An existing value can be changed
student_grades["Alice"] = 90
print(f"\nAfter updating Alice's score: {student_grades}")

# 5. Remove a key:value pair (Optional: for a slightly advanced concept)
del student_grades["Charlie"]
print(f"\nAfter removing Charlie: {student_grades}")
````
