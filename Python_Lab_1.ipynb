## A užduotis

Matavome švytuoklės periodą su chronometru - tris kartus, sekundėmis.

#### A1

Sukurkite tris kintamuosius ir priskirkite jiems matavimų vertes:

`t1 = 2.05`, `t2 = 1.98`, `t3 = 2.11`

Paleiskite langelį. Pastaba: jokio rezultato ekrane **nėra ir neturi būti** - priskyrimas nieko nespausdina.

#### A2

Atspausdinkite `t1` vertę. Tada pabandykite atspausdinti `T1` (didžiąja raide).

Perskaitykite klaidos pranešimą: kaip ji vadinasi ir kuri tai klaidų rūšis?

#### A3

Suskaičiuokite matavimų vidurkį „akivaizdžiu” būdu - **būtent taip, kaip parašyta**:

`vidurkis = t1 + t2 + t3 / 3`

Atspausdinkite rezultatą. Patikrinkite kaip fizikai: visi trys matavimai ≈ 2 s, tai ir vidurkis privalo būti ≈ 2 s. Ar taip ir yra?

#### A4

Programa suveikė be jokio klaidos pranešimo - ir vis tiek **pamelavo**. Kuri tai klaidų rūšis?

Pataisykite formulę skliaustais, kad vidurkis būtų teisingas, ir vėl patikrinkite: ar rezultatas tarp mažiausio ir didžiausio matavimo?

#### A5

Padarykite išvestį gražią: atspausdinkite sakinį su tekstu ir verte, pvz. `Vidutinis periodas: 2.05 s`.

Hint: skaičių suapvalinti galima funkcija `round(vidurkis, 2)` - antras argumentas nurodo skaitmenų kiekį po kablelio. `print()` viduje galima rašyti kelis dalykus, atskirtus kableliais.

#### A6

Laboratorija eksperimentą pakartojo. Nauji duomenys: `t1 = 2.02`, `t2 = 2.09`, `t3 = 1.95`.

Pakeiskite **tik** A1 langelio duomenis ir iš eilės paleiskite žemiau esančius langelius iš naujo. Formulės nelieskite!

Taisyklė, kurią čia atradote: **duomenys atskirai nuo logikos** - pakeitus duomenis, programa lieka ta pati.

#### A7

Chronometras sugedo: ketvirtas matavimas `t4 = 0`.

Pridėkite jį, išplėskite vidurkį iki keturių verčių, ir papildomai suskaičiuokite santykinę sklaidą:

`sklaida = (max(t1, t2, t3, t4) - min(t1, t2, t3, t4)) / min(t1, t2, t3, t4)`

Kas atsitiko ir kodėl? Kuri tai klaidų rūšis? Po eksperimento `t4` ištrinkite ir grąžinkite vidurkį į tris vertes.
