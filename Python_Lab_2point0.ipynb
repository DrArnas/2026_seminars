| 5 | 3 | `15.4842…` |

Paskutiniai skaitmenys gali skirtis - priklauso nuo to, ar kampą skaičiuojate `math.pi / n`, ar `math.radians(180 / n)`. Tai normalu; svarbu pirmieji keturi.
