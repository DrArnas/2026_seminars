**Keturios while ciklo dalys:**

1. **PARUOŠK** kintamąjį, į kurį žiūri sąlyga - **prieš** ciklą
2. **SĄLYGA** - tikrinama **prieš** kiekvieną praėjimą
3. **VEIKSMAI** - ką ciklas daro
4. **PAKEISK** - be jo ciklas nesustos

> **Jei ciklas nesustoja:** Colab - ■ mygtukas šalia langelio (arba Runtime → Interrupt execution, arba Ctrl+M, tada I). Naršyklės neuždarinėkite.

Prie kiekvienos užduoties parašyta, ką turėtumėte pamatyti - **pasitikrinkite patys**, nelaukdami dėstytojo.
Kai kuriose užduotyse kodas jau parašytas, trūksta tik kelių eilučių - jos pažymėtos `____`.

`while` ciklas yra atvirojo tipo (iteracijų skaičiumi neapibrėžtas) ciklas, kartojamas tol kol tenkinama kažkokia sąlyga.

````python
# Sukuriama skaitiklio vertė
count = 1

# While ciklas prieš kiekvieną iteraciją patikrina sąlygą
while count <= 5:
    print(f"Current count is: {count}")

    # Pakeičiam skaitiklio vertę
    count = count + 1

# Ši kodo dalis vykdoma kai while ciklo sąlyga ne(be)tenkinama
print("The loop has finished counting.")
````

Žodynai (dictionaries) kai panaudojami teisingai, yra labai patogus įrankis saugoti informaciją. Informacija saugoma rakto:vertės poromis.

````python
# 1. Sukuriamas žodynas kur raktai yra vardai (strings)
#    o vertės yra testo rezultatai (integers).
student_grades = {"Alice": 88, "Bob": 92, "Charlie": 76}

print("--- Initial Grades ---")
print(student_grades)

# 2. Vertes pasiekiame naudodami raktus (keys)
bobs_score = student_grades["Bob"]
print(f"\nBob's score is: {bobs_score}")

# 3. Galime pridėti naują key:value porą
student_grades["Diana"] = 95
print(f"\nAfter adding Diana: {student_grades}")

# 4. Galime modifikuoti/pakeisti esamą vertę
student_grades["Alice"] = 90
print(f"\nAfter updating Alice's score: {student_grades}")

# 5. Pašalinti key:value porą (Optional: for a slightly advanced concept)
del student_grades["Charlie"]
print(f"\nAfter removing Charlie: {student_grades}")
````

#### B1 - Kiek laiko kamuoliukas šokinėja?

Laisvai krentant iš aukščio $h$, kritimo laikas $t = \sqrt{2h/g}$ (2 savaitės bokšto formulė). Vienas atšokimas iki aukščio $h$ trunka $2\sqrt{2h/g}$ (aukštyn ir žemyn).

Su `while` sumuokite atšokimų laikus (`g = 9.81`, daugiklis `0.8`), **kol** vieno atšokimo laikas tampa mažesnis už $10^{-6}$ s. Pradėkite nuo pirmo kritimo laiko.
Atspausdinkite bendrą laiką ir kiek atšokimų prireikė.

*Turi pasirodyti:* `11.4583 s`, `126` atšokimai

Palyginkite su tikslia formule $t_0\left(1 + \dfrac{2\sqrt{0{,}8}}{1-\sqrt{0{,}8}}\right)$, kur $t_0 = \sqrt{2\cdot 2{,}0/g}$ - turi sutapti iki ~$10^{-5}$ s. Nepamirškite antrosios sąlygos!

#### B2 - Visi kamuoliukai viename paveiksle

Kodas paruoštas: jis nubraižo visų keturių kamuoliukų aukščius viename paveiksle. Dvi naujos detalės:
`label=rusis` kiekvienai kreivei suteikia vardą, o `plt.legend()` parodo vardų lentelę.
Atkreipkite dėmesį: `while` ciklas čia yra **for ciklo viduje** - kiekvienam kamuoliukui atskirai.

Užrašykite ašių pavadinimus ir paleiskite.

*Turi pasirodyti:* keturios kreivės su legenda. Komentare atsakykite: kuris kamuoliukas nurimsta greičiausiai ir kaip tai matyti paveiksle?

````python
for rusis, verte in kamuoliukai.items():
    numeris = []
    aukstis = []
    h = 2.0
    n = 0
    while h >= 0.01:
        h = h * verte
        n = n + 1
        numeris.append(n)
        aukstis.append(h)
    plt.plot(numeris, aukstis, "o-", label=rusis)

plt.xlabel("____")
plt.ylabel("____")
plt.legend()
plt.show()
````
