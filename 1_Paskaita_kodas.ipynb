```python
# Load the data — one line
import pandas as pd

data = pd.read_csv("global_temperature.csv", comment="#")
data
```

*176 years of global measurements, loaded in one line. Each number is how much warmer or cooler that year was than the 20th-century average.

```python
# Which year was the hottest?
hottest = data.loc[data["anomaly_C"].idxmax()]
hottest
```

```python
# How many years were more than half a degree above average?
warm = data[data["anomaly_C"] > 0.5]
print("Years above +0.5 °C:", len(warm))
print("The first of them:", warm["year"].min())
```

```python
# Average of the last 10 years, in one line
round(data["anomaly_C"].tail(10).mean(), 2)
```

```python
# The picture that says it all
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(data["year"], data["anomaly_C"], color="grey", lw=1)
ax.scatter(warm["year"], warm["anomaly_C"], color="crimson", s=14,
           label="more than +0.5 °C")
ax.axhline(0, color="black", lw=0.8)
ax.set_xlabel("Year")
ax.set_ylabel("Temperature anomaly (°C)")
ax.set_title("Global mean temperature, 1850–2025")
ax.legend()
plt.show()
```
